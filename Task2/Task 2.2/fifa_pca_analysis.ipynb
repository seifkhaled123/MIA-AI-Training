# FIFA 19 player performance with PCA

This notebook runs two PCA implementations without using a pre-built PCA library. The numerical features are standardized, a covariance matrix is calculated, its eigenpairs are sorted, and the data is projected onto PC1 and PC2. The original `Position` column is retained only for visualization.

```python
from IPython.display import display, Image
from pca_analysis import run_analysis

analysis = run_analysis()
print(f"Players analyzed: {len(analysis['players']):,}")
```

## Explained variance and component loadings

```python
display(analysis['explained_variance'].round(4))
display(analysis['loadings'].round(4))
```

## Players projected onto PC1 and PC2

```python
display(Image(filename=str(analysis['plot_path'])))
```

## NumPy versus pure Python

```python
benchmark = analysis['benchmark']
print(f"NumPy median time:       {benchmark['numpy_seconds']:.6f} s")
print(f"Pure Python median time: {benchmark['pure_python_seconds']:.6f} s")
print(f"NumPy speed-up:          {benchmark['speedup']:.1f}x")
print(f"Maximum score error:     {benchmark['max_score_absolute_error']:.3e}")
print(f"Maximum covariance error:{benchmark['max_covariance_absolute_error']:.3e}")
```
